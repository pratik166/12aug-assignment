
#### Take the shopping amount as input.

#### If amount is 5000 or more → give 20% discount
#### If amount is 3000 or more → give 10% discount
#### Otherwise → no discount

#### Print:

#### Original amount
#### Discount amount
#### Final amount after discount

```python
amount = float(input("Enter shopping amount: "))

if amount >= 5000:
    discount_percentage = 20
    discount = amount * discount_percentage / 100

elif amount >= 3000:
    discount_percentage = 10
    discount = amount * discount_percentage / 100

else:
    discount_percentage = 0
    discount = 0

final_amount = amount - discount

print("Original Amount:", amount)
print("Discount Percentage:", discount_percentage, "%")
print("Discount Amount:", discount)
print("Final Amount:", final_amount)
```

#### To find year is leap year or not

```python
year = int(input("Enter a year: "))

if year % 400 == 0:
    print("LEAP YEAR")
elif year % 4 == 0 and year % 100 != 0:
    print("LEAP YEAR")
else:
    print("NOT A LEAP YEAR")
```
